## Άσκηση 16
Δημιουργήστε μια κλάση `BankAccount` (Τραπεζικός Λογαριασμός) με μεθόδους `deposit` (κατάθεση) και `withdraw` (ανάληψη) που να ενημερώνουν το υπόλοιπο.

```python
class BankAccount:
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self.balance = initial_balance
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"Deposited ${amount:.2f}"
        return "Deposit amount must be positive"
    
    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            return f"Withdrew ${amount:.2f}"
        return "Insufficient funds or invalid amount"
    
    def balance_check(self):
        return self.balance   
    
    def __str__(self):
        """Safe representation - only shows current state"""
        return (f"Account Owner : {self.owner}\n"
                f"Current Balance: ${self.balance:.2f}")

call = BankAccount("Bob", 1000)
print(call.deposit(600))
print(call.balance_check())
print(call.owner)
```

## Άσκηση 18
Γράψτε ένα πρόγραμμα που δείχνει τη χρήση της μαγικής μεθόδου `__str__` ή `__repr__` για να αναπαραστήσει ως κείμενο τα δεδομένα ενός αντικειμένου της επιλογής σας.

```python
class BankAccount:


    def __init__(self, owner, initial_balance=0):

        self.owner = owner
        self.balance = initial_balance
    
    def deposit(self, amount):

        if amount > 0:
            self.balance += amount
            return f"Deposited ${amount:.2f}"
        return "Deposit amount must be positive"
    
    def withdraw(self, amount):

        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            return f"Withdrew ${amount:.2f}"
        return "Insufficient funds or invalid amount"
    
    def balanceCheck(self):

        return self.balance   
    
    def __str__(self):
        """Provides a user-friendly text representation of the account."""
        return (f"Account Owner : {self.owner}\n"
                f"Current Balance: ${self.balance:.2f}")

    def __repr__(self):
        """Provides an unambiguous developer representation of the account."""
        return f"BankAccount(owner='{self.owner}', initial_balance={self.balance})"



account = BankAccount("Bob", 1000)

print("--- Operations ---")
print(account.deposit(600))
print(f"Owner: {account.owner}")
print(f"Checked Balance: ${account.balanceCheck():.2f}")

print("\n--- Magic Methods Demonstration ---")
# 1. Demonstrating __str__ (User-facing text representation)
print("Using __str__ via print(account):")
print(account)

print("\nUsing __repr__ via print(repr(account)):")
# 2. Demonstrating __repr__ (Developer-facing representation)
print(repr(account))
```

## Άσκηση 20
Γράψτε έναν decorator που να τυπώνει τον χρόνο εκτέλεσης (χρησιμοποιήστε το module time) οποιασδήποτε συνάρτησης στην οποία εφαρμόζεται.

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper


@timer
def slow_function():
    time.sleep(1)
    return "Done!"

slow_function()
```
