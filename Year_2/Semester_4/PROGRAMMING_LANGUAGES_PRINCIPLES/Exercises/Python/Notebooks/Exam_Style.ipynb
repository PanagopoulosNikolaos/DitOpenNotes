# Εξάσκηση: Ερωτήσεις Προσομοίωσης Εξετάσεων
Απαντήστε σε κάθε ερώτηση γράφοντας ένα πρόγραμμα Python που λύνει το συγκεκριμένο πρόβλημα. Γράψτε την απάντησή σας στο κενό κελί κώδικα κάτω από κάθε ερώτημα.


## Άσκηση 1
Δίνεται ο κώδικας: `x = 5\ndef foo():\n    x = 10\n    return x`. Τι θα επιστρέψει η `foo()` και ποια η τιμή του global `x` μετά; Γράψτε το πρόγραμμα και τυπώστε τα αποτελέσματα για να εξηγήσετε την τοπική έναντι καθολικής εμβέλειας.

## Άσκηση 2
Τι θα προκύψει από την εκτέλεση: `{w: len(w) for w in ['exam', 'python', 'code']}`; Γράψτε τον κώδικα για να το ελέγξετε.

## Άσκηση 3
Υλοποιήστε το list comprehension `[num for num in range(10) if num % 2 == 0]` με απλό βρόχο `for` και συγκρίνετε τα αποτελέσματα τυπώνοντάς τα.

## Άσκηση 4
Γράψτε μια συνάρτηση που μετατρέπει βαθμούς Κελσίου σε Φαρενάιτ. Διαχειριστείτε ορθά την περίπτωση μη-αριθμητικής εισόδου χρησιμοποιώντας `try/except` για `TypeError` ή ελέγχοντας τον τύπο.

## Άσκηση 5
Γράψτε κώδικα Python που λαμβάνει μια λίστα από λίστες (πίνακα 2x3 πχ. `[[1,2,3],[4,5,6]]`) και την 'επιπεδώνει' (flatten) σε μια μονοδιάστατη λίστα.

## Άσκηση 6
Εξηγήστε και επιβεβαιώστε με κώδικα τη συμπεριφορά των μεταβλητών κλάσης (class variables) έναντι των μεταβλητών στιγμιότυπου (instance variables) χρησιμοποιώντας μια κλάση `Car`.

## Άσκηση 7
Τι κάνει η έκφραση `list(map(lambda x: x*2, filter(lambda x: x>0, [-1, 0, 1, 2])))`; Γράψτε τον κώδικα και στη συνέχεια γράψτε το αντίστοιχο σε list comprehension.

## Άσκηση 8
Γράψτε ένα πρόγραμμα που υλοποιεί τον υπολογισμό της ν-οστής δύναμης ενός αριθμού (x^n) αναδρομικά.

## Άσκηση 9
Δημιουργήστε μια αφηρημένη κλάση `Shape` (χρησιμοποιώντας το module `abc` ή απλά υψώνοντας NotImplementedError) με μια μέθοδο `area()`. Μετά, δημιουργήστε `Circle` και `Square` που κληρονομούν την `Shape` και υλοποιούν την `area()`.

## Άσκηση 10
Γράψτε μια συνάρτηση `merge_dicts(d1, d2)` που δέχεται δύο λεξικά με αριθμητικές τιμές και τα συγχωνεύει. Σε περίπτωση κοινού κλειδιού, οι τιμές να προστίθενται.

## Άσκηση 11
Γράψτε κώδικα που δημιουργεί 2 διαφορετικά αντικείμενα λίστας με τα ίδια στοιχεία. Χρησιμοποιήστε τους τελεστές `is` και `==` για να δείξετε έμπρακτα τη διαφορά μεταξύ ταυτότητας και ισότητας.

## Άσκηση 12
Τι τυπώνει η εντολή `print('python'[::-1])`; Δικαιολογήστε το μέσω επίδειξης slicing με διάφορα βήματα (steps).

## Άσκηση 13
Γράψτε έναν generator `primes(n)` που παράγει με τη σειρά όλους τους πρώτους αριθμούς μέχρι τον αριθμό n.

## Άσκηση 14
Γράψτε ένα script που χρησιμοποιεί το module `re` (regular expressions) για να εντοπίσει αν ένα δοσμένο string έχει τη σωστή μορφή email (π.χ. περιέχει @ και .).

## Άσκηση 15
Χρησιμοποιώντας το module `sqlite3`, δημιουργήστε μια in-memory βάση δεδομένων, έναν πίνακα `students` (με πεδία id, name) και εισάγετε δύο εγγραφές. Έπειτα τυπώστε τες.

## Άσκηση 16
Γράψτε ένα πρόγραμμα που δείχνει τη διαφορά μεταξύ τοπικής (Local), περικλείουσας (Enclosing) και καθολικής (Global) εμβέλειας (κανόνας LEGB) στην Python.

## Άσκηση 17
Δημιουργήστε έναν decorator `@logger` που θα μπορεί να εφαρμοστεί σε οποιαδήποτε συνάρτηση για να τυπώνει τα ορίσματα (args, kwargs) με τα οποία κλήθηκε η συνάρτηση πριν την εκτέλεσή της.

## Άσκηση 18
Υλοποιήστε με κώδικα το Singleton design pattern στην Python (μια κλάση που επιτρέπει να δημιουργηθεί μόνο ένα instance της) ελέγχοντας τη μέθοδο `__new__`.

## Άσκηση 19
Δείξτε τη χρήση του `yield from` στην Python δημιουργώντας έναν generator που παράγει στοιχεία αναθέτοντας τη δουλειά σε έναν άλλον generator.

## Άσκηση 20
Γράψτε μια συνάρτηση που δέχεται αυθαίρετο αριθμό keyword arguments (`**kwargs`) και τυπώνει μόνο τα κλειδιά των οποίων η τιμή είναι τύπου string.
